```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as pyplt
```

```python
vm = pd.read_csv("../IO-VNBD/Synchronised V abd S datasets/Categorised IOVNB Dataset/M (Driver B)/V-M.csv")
vm.columns = vm.columns.str.strip()
```

```python
ms = pd.read_csv("../IO-VNBD/Synchronised V abd S datasets/Categorised IOVNB Dataset/M (Driver B)/S-M.csv", encoding="cp1252")
ms.columns = ms.columns.str.strip()
```

```python
# Check whether the synchronized files have matching row counts

print("Vehicle rows:", len(vm))
print("Sensor rows:", len(ms))

print("\nFirst 10 vehicle velocities:")
print(vm["Velocity (km/hr)"].head(10).to_list())

print("\nFirst 10 GPS speeds:")
print(ms["GPS SPEED (Kmh)"].head(10).to_list())
```

```python
# Features for CNN
features = [
    "ACCELEROMETER X (m/s²)",
    "ACCELEROMETER Y (m/s²)",
    "ACCELEROMETER Z (m/s²)",
    "GYROSCOPE Yaw (rad/s)",
    "GYROSCOPE Pitch (rad/s)",
    "GYROSCOPE Roll (rad/s)"
]

X_raw = ms[features].to_numpy(dtype=np.float32)

# Label from vehicle velocity
# 1 = stationary
# 0 = moving
y_raw = (vm["Velocity (km/hr)"] < 1.0).astype(np.int64).to_numpy()

# Make lengths equal
n = min(len(X_raw), len(y_raw))

X_raw = X_raw[:n]
y_raw = y_raw[:n]

print("X_raw:", X_raw.shape)
print("y_raw:", y_raw.shape)
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

print("TensorFlow:", tf.__version__)
print("GPU devices:", tf.config.list_physical_devices("GPU"))
```

```python
# Create leakage-safe time windows and normalize every training target
speed_raw = vm["Velocity (km/hr)"].to_numpy(dtype=np.float32)[:n]

window_size = 50   # 5 seconds at the dataset's 0.1 second sample period
stride = 5         # one training example every 0.5 seconds
train_end = int(n * 0.70)
validation_end = int(n * 0.85)

train_mean = X_raw[:train_end].mean(axis=0)
train_std = X_raw[:train_end].std(axis=0)
train_std = np.where(train_std < 1e-6, 1.0, train_std)
X_scaled = (X_raw - train_mean) / train_std

speed_mean = float(speed_raw[:train_end].mean())
speed_std = max(float(speed_raw[:train_end].std()), 1e-6)
speed_scaled = (speed_raw - speed_mean) / speed_std


def make_windows(start, end):
    window_data = []
    stationary_targets = []
    speed_targets = []

    last_start = end - window_size + 1
    for window_start in range(start, last_start, stride):
        window_end = window_start + window_size
        window_data.append(X_scaled[window_start:window_end])

        # Require most of a window to be stationary to reduce transition noise.
        stationary_targets.append(
            float(y_raw[window_start:window_end].mean() >= 0.8)
        )
        speed_targets.append(float(np.median(speed_scaled[window_start:window_end])))

    return (
        np.asarray(window_data, dtype=np.float32),
        np.asarray(stationary_targets, dtype=np.float32),
        np.asarray(speed_targets, dtype=np.float32),
    )


X_train, y_train_stationary, y_train_speed = make_windows(0, train_end)
X_validation, y_validation_stationary, y_validation_speed = make_windows(
    train_end, validation_end
)
X_test, y_test_stationary, y_test_speed = make_windows(validation_end, n)

print("Train:", X_train.shape)
print("Validation:", X_validation.shape)
print("Test:", X_test.shape)
print("Stationary ratio:", y_train_stationary.mean())
print(f"Speed normalization: mean={speed_mean:.3f}, std={speed_std:.3f}")
```

```python
# Regularized residual-style 1D CNN with stationary and speed heads
sensor_input = keras.Input(
    shape=(window_size, len(features)),
    name="sensor_window",
)

x = layers.GaussianNoise(0.02)(sensor_input)
x = layers.Conv1D(64, kernel_size=7, padding="same", use_bias=False)(x)
x = layers.BatchNormalization()(x)
x = layers.Activation("relu")(x)
x = layers.MaxPooling1D(pool_size=2)(x)
x = layers.SpatialDropout1D(0.15)(x)

residual = layers.Conv1D(128, kernel_size=1, padding="same", use_bias=False)(x)
x = layers.Conv1D(128, kernel_size=5, padding="same", use_bias=False)(x)
x = layers.BatchNormalization()(x)
x = layers.Activation("relu")(x)
x = layers.Conv1D(128, kernel_size=5, padding="same", use_bias=False)(x)
x = layers.BatchNormalization()(x)
x = layers.Add()([x, residual])
x = layers.Activation("relu")(x)
x = layers.MaxPooling1D(pool_size=2)(x)
x = layers.SpatialDropout1D(0.20)(x)

x = layers.Conv1D(192, kernel_size=3, padding="same", activation="relu")(x)
x = layers.BatchNormalization()(x)
x = layers.GlobalAveragePooling1D()(x)
x = layers.Dense(96, activation="relu", kernel_regularizer=keras.regularizers.l2(1e-4))(x)
x = layers.Dropout(0.30)(x)

stationary_output = layers.Dense(
    1,
    activation="sigmoid",
    name="stationary_probability",
)(x)
speed_output = layers.Dense(1, name="speed_kmh")(x)

cnn_model = keras.Model(
    inputs=sensor_input,
    outputs={
        "stationary_probability": stationary_output,
        "speed_kmh": speed_output,
    },
)

cnn_model.compile(
    optimizer=keras.optimizers.AdamW(learning_rate=3e-4, weight_decay=1e-4),
    loss={
        "stationary_probability": keras.losses.BinaryCrossentropy(),
        "speed_kmh": keras.losses.Huber(delta=1.0),
    },
    loss_weights={
        "stationary_probability": 0.7,
        "speed_kmh": 0.3,
    },
    metrics={
        "stationary_probability": [
            keras.metrics.BinaryAccuracy(name="accuracy"),
            keras.metrics.AUC(name="auc"),
            keras.metrics.Precision(name="precision"),
            keras.metrics.Recall(name="recall"),
        ],
        "speed_kmh": [keras.metrics.MeanAbsoluteError(name="mae")],
    },
)

cnn_model.summary()
```

```python
# Balance the stationary classifier and train with early stopping
positive_count = max(float(y_train_stationary.sum()), 1.0)
negative_count = max(float(len(y_train_stationary) - y_train_stationary.sum()), 1.0)
positive_weight = np.sqrt(negative_count / positive_count)
stationary_sample_weight = np.where(
    y_train_stationary == 1.0,
    positive_weight,
    1.0,
).astype(np.float32)

callbacks = [
    keras.callbacks.EarlyStopping(
        monitor="val_stationary_probability_auc",
        mode="max",
        patience=10,
        restore_best_weights=True,
    ),
    keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        mode="min",
        factor=0.5,
        patience=4,
        min_lr=1e-6,
    ),
]

history = cnn_model.fit(
    X_train,
    {
        "stationary_probability": y_train_stationary,
        "speed_kmh": y_train_speed,
    },
    sample_weight={
        "stationary_probability": stationary_sample_weight,
        "speed_kmh": np.ones_like(y_train_speed),
    },
    validation_data=(
        X_validation,
        {
            "stationary_probability": y_validation_stationary,
            "speed_kmh": y_validation_speed,
        },
    ),
    epochs=80,
    batch_size=256,
    callbacks=callbacks,
    verbose=1,
)
```

```python
# Evaluate both CNN outputs and calibrate the ZUPT threshold on validation data
validation_predictions = cnn_model.predict(X_validation, verbose=0)
validation_probability = validation_predictions["stationary_probability"].reshape(-1)

thresholds = np.linspace(0.10, 0.90, 81)
f1_scores = []
for threshold in thresholds:
    predicted = (validation_probability >= threshold).astype(np.int64)
    true_positive = np.sum((predicted == 1) & (y_validation_stationary == 1))
    false_positive = np.sum((predicted == 1) & (y_validation_stationary == 0))
    false_negative = np.sum((predicted == 0) & (y_validation_stationary == 1))
    precision_value = true_positive / max(true_positive + false_positive, 1)
    recall_value = true_positive / max(true_positive + false_negative, 1)
    f1_scores.append(
        2 * precision_value * recall_value / max(precision_value + recall_value, 1e-8)
    )

zupt_threshold = float(thresholds[int(np.argmax(f1_scores))])

test_results = cnn_model.evaluate(
    X_test,
    {
        "stationary_probability": y_test_stationary,
        "speed_kmh": y_test_speed,
    },
    return_dict=True,
    verbose=0,
)

predictions = cnn_model.predict(X_test, verbose=0)
predicted_stationary = (
    predictions["stationary_probability"].reshape(-1) >= zupt_threshold
).astype(np.int64)

actual_stationary = y_test_stationary.astype(np.int64)
true_positive = np.sum((predicted_stationary == 1) & (actual_stationary == 1))
false_positive = np.sum((predicted_stationary == 1) & (actual_stationary == 0))
false_negative = np.sum((predicted_stationary == 0) & (actual_stationary == 1))

precision = true_positive / max(true_positive + false_positive, 1)
recall = true_positive / max(true_positive + false_negative, 1)
accuracy = np.mean(predicted_stationary == actual_stationary)
predicted_speed = predictions["speed_kmh"].reshape(-1) * speed_std + speed_mean
actual_speed = y_test_speed * speed_std + speed_mean
speed_mae = np.mean(np.abs(predicted_speed - actual_speed))

print("Test metrics:", test_results)
print(f"Calibrated ZUPT threshold: {zupt_threshold:.2f}")
print(f"Stationary accuracy: {accuracy:.4f}")
print(f"Stationary precision: {precision:.4f}")
print(f"Stationary recall: {recall:.4f}")
print(f"Speed MAE (km/hr): {speed_mae:.4f}")
```

```python
# Save the trained model and preprocessing parameters for INS + EKF integration
cnn_model.save("cnn_zupt_speed.keras")
np.savez(
    "cnn_preprocessing.npz",
    mean=train_mean,
    std=train_std,
    speed_mean=np.asarray(speed_mean),
    speed_std=np.asarray(speed_std),
    zupt_threshold=np.asarray(zupt_threshold),
    features=np.asarray(features),
    window_size=np.asarray(window_size),
)
print("Saved cnn_zupt_speed.keras and cnn_preprocessing.npz")
```

```python
def predict_sensor_window(sensor_window):
    """Return calibrated ZUPT probability and speed in km/hr."""
    sensor_window = np.asarray(sensor_window, dtype=np.float32)
    if sensor_window.shape != (window_size, len(features)):
        raise ValueError(
            f"Expected shape {(window_size, len(features))}, "
            f"got {sensor_window.shape}"
        )

    normalized_window = (sensor_window - train_mean) / train_std
    model_input = normalized_window[np.newaxis, ...]
    output = cnn_model.predict(model_input, verbose=0)
    stationary_probability = float(output["stationary_probability"][0, 0])
    speed_kmh = float(output["speed_kmh"][0, 0] * speed_std + speed_mean)

    return {
        "stationary_probability": stationary_probability,
        "zupt_decision": stationary_probability >= zupt_threshold,
        "speed_kmh": speed_kmh,
    }

print(predict_sensor_window(X_raw[:window_size]))
```

```python
# Write a reproducible text report with the final held-out test results
true_negative = np.sum((predicted_stationary == 0) & (actual_stationary == 0))
f1 = 2 * precision * recall / max(precision + recall, 1e-8)
specificity = true_negative / max(true_negative + false_positive, 1)
balanced_accuracy = (recall + specificity) / 2.0
speed_rmse = float(np.sqrt(np.mean((predicted_speed - actual_speed) ** 2)))
speed_r2 = float(
    1.0
    - np.sum((actual_speed - predicted_speed) ** 2)
    / max(np.sum((actual_speed - actual_speed.mean()) ** 2), 1e-8)
)
best_epoch = int(np.argmax(history.history["val_stationary_probability_auc"]) + 1)

report = f"""CNN TEST REPORT
===============
Generated: {pd.Timestamp.now()}

DATA
----
Source rows: {n}
Training windows: {len(X_train)}
Validation windows: {len(X_validation)}
Test windows: {len(X_test)}
Window size: {window_size} samples
Window stride: {stride} samples
Input features: {", ".join(features)}

TRAINING
--------
Epochs completed: {len(history.history["loss"])}
Best validation AUC epoch: {best_epoch}
Stationary training ratio: {y_train_stationary.mean():.6f}
Speed mean (km/hr): {speed_mean:.6f}
Speed standard deviation: {speed_std:.6f}

STATIONARY / ZUPT TEST METRICS
------------------------------
Accuracy: {accuracy:.6f}
AUC: {test_results["stationary_probability_auc"]:.6f}
Precision: {precision:.6f}
Recall / Sensitivity: {recall:.6f}
F1 score: {f1:.6f}
Specificity: {specificity:.6f}
Balanced accuracy: {balanced_accuracy:.6f}
ZUPT threshold: {zupt_threshold:.6f}
True positives: {int(true_positive)}
True negatives: {int(true_negative)}
False positives: {int(false_positive)}
False negatives: {int(false_negative)}

SPEED TEST METRICS
------------------
MAE (km/hr): {speed_mae:.6f}
RMSE (km/hr): {speed_rmse:.6f}
R2: {speed_r2:.6f}

ARTIFACTS
---------
Model: cnn_zupt_speed.keras
Preprocessing: cnn_preprocessing.npz
"""

with open("cnn_test_report.txt", "w", encoding="utf-8") as report_file:
    report_file.write(report)

print(report)
print("Saved cnn_test_report.txt")
```
